```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):

    def sound(self):
        print("Dog barks")

d1 = Dog()
d1.sound()
```

```python
# Single Inheritance
class Father:
    def show_father(self):
        print("This is Father")

class Son(Father):
    def show_son(self):
        print("This is Son")

s1 = Son()
s1.show_father()
s1.show_son()
```

```python
# Multilevel Inheritance
class Grandfather:
    def show_grandfather(self):
        print("This is Grandfather")

class Father(Grandfather):
    def show_father(self):
        print("This is Father")

class Son(Father):
    def show_son(self):
        print("This is Son")

s1 = Son()
s1.show_grandfather()
s1.show_father()
s1.show_son()
```

```python
# Multiple Inheritance
class Father:
    def show_father(self):
        print("This is Father")

class Mother:
    def show_mother(self):
        print("This is Mother")

class Child(Father, Mother):
    def show_child(self):
        print("This is Child")

c1 = Child()
c1.show_father()
c1.show_mother()
c1.show_child()
```

```python
# Hierarchical Inheritance
class Animal:
    def eat(self):
        print("Animal can eat")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")

d1 = Dog()
c1 = Cat()

d1.eat()
d1.bark()

c1.eat()
c1.meow()
```

```python
# Hybrid Inheritance
class Person:
    def show_person(self):
        print("This is Person")

class Student(Person):
    def show_student(self):
        print("This is Student")

class Employee(Person):
    def show_employee(self):
        print("This is Employee")

class Intern(Student, Employee):
    def show_intern(self):
        print("This is Intern")

i1 = Intern()
i1.show_person()
i1.show_student()
i1.show_employee()
i1.show_intern()
```
