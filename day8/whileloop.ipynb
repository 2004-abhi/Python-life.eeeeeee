```python
# no is prime or not
num = int(input("Enter a number: "))

i = 2
is_prime = True

while i < num:
    if num % i == 0:
        is_prime = False
        break
    i += 1

if num <= 1:
    print("Not Prime")

elif is_prime:
    print("Prime")

else:
    print("Not Prime")
```

```python
# Take a range of number from 1 to 100, print first no is div by 3 not by 5 ,second num div by 3 and 5, third neither by 3 and 5
i = 1

while i <= 100:

    if i % 3 == 0 and i % 5 != 0:
        print(i, "divisible by 3 but not by 5")

    elif i % 3 == 0 and i % 5 == 0:
        print(i, "divisible by both 3 and 5")

    elif i % 3 != 0 and i % 5 != 0:
        print(i, "neither divisible by 3 nor 5")

    else:
        print(i, "divisible by 5 but not by 3")

    i += 1
```

```python
#  prime or not 
n = int(input("Enter a number: "))

i = 1
count = 0

while i <= n:
    if n % i == 0:
        count += 1
    i += 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```
