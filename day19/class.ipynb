```python
# 1. Static Attribute (Class Attribute)
# Definition: A static attribute is a variable that belongs to the class and can be shared by all its objects. 
# In Python, it is commonly called a class attribute.
class Student:
    college = "MIT"  # Static attribute

s1 = Student()
s2 = Student()

print(s1.college)
print(s2.college)
print(Student.college)
# college is declared directly inside the class.
# Both objects can access the same class attribute.
# We can access it through the class name or an object
```

```python
# 2. Non-static Attribute (Instance Attribute)
# Definition: A non-static attribute is a variable that belongs to a particular object. 
# In Python, it is commonly called an instance attribute.
class Student:
    def __init__(self):
        self.name = "Tom"
        self.age = 22

s1 = Student()

print(s1.name)
print(s1.age)
# self.name and self.age are instance attributes.
# self refers to the current object.
# Each object can have its own values
```

```python
# 3. Constructor (__init__)
# Definition: A constructor is commonly described as a special method that runs automatically when an object is created.
#  In Python, __init__() initializes the object's attributes.
# class calling itself is called constructor
class Student:
    def __init__(self):
        print("Student object initialized")

s1 = Student()

# Student is the class.
# s1 = Student() creates an object.
# Python automatically calls __init__().
# The message is printed.
```

```python
# 4. Default Constructor
# Definition: A default constructor usually refers to creating an object without passing any arguments.
#  If you don't define your own __init__() method, Python provides a default initializer.
class Student:
    pass

s1 = Student()

print("Object created")
# pass means the class has no statements to execute.
# Student() creates an object without passing any arguments.
# Python supplies the default initialization behavior.
```

```python
# 5. What is a parameterized constructor?
# A parameterized constructor is an initializer that accepts parameters 
# so we can give different values to different objects when creating them.
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Tom", 22)
s2 = Student("Sam", 20)

print(s1.name)
print(s1.age)

print(s2.name)
print(s2.age)
# 
```

```python
class employee:
    def __init__(self,emp_id,emp_name,emp_sal):
        self.emp_id=emp_id
        self.emp_name=emp_name
        self.emp_sal=emp_sal

    def show_details(self):
        print(self.emp_id,self.emp_name,self.emp_sal)

    def bonus_sal(self):  
        if self.emp_sal>20000 and self.emp_sal<30000 :
            bonus_sal=self.emp_sal + (self.emp_sal*0.15)
        self.show_details()
        print(bonus_sal)

emp1=employee(1,"abhi",25000)  
emp1.bonus_sal()     
print("-------------------------------------------")
emp2=employee(2,"arya",24000)
emp2.bonus_sal()
```
