```python
# 1. Static Attribute (Class Attribute)
# Definition: A static attribute is a variable that belongs to the class and can be shared by all its objects. 
# In Python, it is commonly called a class attribute.
class Student:
    college = "MIT"  # Static attribute

s1 = Student()
s2 = Student()

print(s1.college)
print(s2.college)
print(Student.college)
# college is declared directly inside the class.
# Both objects can access the same class attribute.
# We can access it through the class name or an object
```

```python
# 2. Non-static Attribute (Instance Attribute)
# Definition: A non-static attribute is a variable that belongs to a particular object. 
# In Python, it is commonly called an instance attribute.
class Student:
    def __init__(self):
        self.name = "Tom"
        self.age = 22

s1 = Student()

print(s1.name)
print(s1.age)
# self.name and self.age are instance attributes.
# self refers to the current object.
# Each object can have its own values
```

```python
# 3. Constructor (__init__)
# Definition: A constructor is commonly described as a special method that runs automatically when an object is created.
#  In Python, __init__() initializes the object's attributes.
# class calling itself is called constructor
class Student:
    def __init__(self):
        print("Student object initialized")

s1 = Student()

# Student is the class.
# s1 = Student() creates an object.
# Python automatically calls __init__().
# The message is printed.
```

```python
# 4. Default Constructor
# Definition: A default constructor usually refers to creating an object without passing any arguments.
#  If you don't define your own __init__() method, Python provides a default initializer.
class Student:
    pass

s1 = Student()

print("Object created")
# pass means the class has no statements to execute.
# Student() creates an object without passing any arguments.
# Python supplies the default initialization behavior.
```

```python
class employee:
    def __init__(self,emp_id,emp_name,emp_sal):
        self.emp_id=emp_id
        self.emp_name=emp_name
        self.emp_sal=emp_sal

    def show_details(self):
        print(self.emp_id,self.emp_name,self.emp_sal)

    def bonus_sal(self):  
        if self.emp_sal>20000 and self.emp_sal<30000 :
            bonus_sal=self.emp_sal + (self.emp_sal*0.15)
        self.show_details()
        print(bonus_sal)

emp1=employee(1,"abhi",25000)  
emp1.bonus_sal()     
print("-------------------------------------------")
emp2=employee(2,"arya",24000)
emp2.bonus_sal()
```

```python
# Q6. Rectangle Calculation
# Create a class Rectangle with a constructor that accepts length and breadth. Create methods to calculate and display the area and perimeter.
class rectangle:
    def __init__(self,length,breadh):
        self.length=length
        self.breadth=breadh
    def calculator(self):
        area = self.length * self.breadth
        perimeter = 2 * (self.length + self.breadth)

        print("Area:", area)
        print("Perimeter:", perimeter)

r1=rectangle(10,5)
r1.calculator()

    
```

```python
# Q7. Circle Calculation
# Create a class Circle with a constructor that accepts radius. Create methods to calculate the area and circumference.
class circle:
    def __init__(self,radius):
        self.radius=radius
    def calculate(self):
        area=3.14*self.radius*self.radius
        circumference=2*3.14*self.radius

        print("area of circle is :",area)
        print("circumference of circle is:",circumference)

c1=circle(3)
c1.calculate()        

    
```

```python
# Q8. Bank Account
# Create a class BankAccount with a constructor that accepts account_holder and balance. Create methods to:
# Display account details.
# Deposit an amount.
# Display the updated balance.

class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance

    def account_details(self):
        print("account holder :", self.account_holder)
        print("balance :", self.balance)

    def deposit(self,ammount):  
        self.balance = self.balance + ammount

    def display_balance(self):
        print("Updated Balance:", self.balance)
    
a1=BankAccount("abhinand",50000)

a1.account_details()
a1.deposit(500)
a1.display_balance()

        
```

```python
# Q9. Student Marks
# Create a class Student with a constructor that accepts name, mark1, mark2, and mark3. Create methods to calculate the total and average marks.
class student:
    def __init__(self,name,mark1,mark2,mark3):
        self.name=name
        self.mark1=mark1
        self.mark2=mark2
        self.mark3=mark3

    def total(self):
        total = self.mark1 + self.mark2 + self.mark3
        print("student total mark :", total)

    def average(self):
        avg = (self.mark1 + self.mark2 + self.mark3) / 3
        print("Average Marks:", avg)

s1=student("Abu",29,40,38)
s1.total()
s1.average()


        
```

```python
# Q10. Product Discount
# Create a class Product with a constructor that accepts name and price.
#  Create a method to calculate a 10% discount and another method to display the final price.
 
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price

    def calculator(self)  : 
        self.discount = self.price * 0.10
        self.final_price = self.price - self.discount

    def display(self)  :
        print("product name :",self.name)
        print("product price :",self.price)
        print("discount of product :", self.discount)
        print("final price ,", self.final_price)  

d1=product("samsung s26 ultra",1300000)
d1.calculator()
d1.display()

        
```

```python
# Q18. Constructor and Multiple Operations
# Create a class BankAccount with a constructor that accepts name and balance. Create methods to:
# Deposit money.
# Withdraw money, provided the balance is sufficient.
# Display the current balance.


class BankAccount:
    def __init__(self, Name, Balance):
        self.Name = Name
        self.Balance = Balance

    def Deposit(self, amount):
        self.Balance = self.Balance + amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if self.Balance >= amount:
            self.Balance = self.Balance - amount
            print("Withdraw amount:", amount)
        else:
            print("Insufficient amount")

    def display_balance(self):
        print("Account Holder:", self.Name)
        print("Current Balance:", self.Balance)


a1 = BankAccount("Abhinand", 10000)

a1.Deposit(2000)
a1.withdraw(3000)
a1.display_balance()
```

```python
# Q19. Compare Two Objects
# Create a class Product with a constructor that accepts name and price.
#  Create two objects and a method that compares their prices and displays which product is more expensive.
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def compare(self, p):
        if self.price > p.price:
            print(self.name, "is more expensive")
        elif self.price < p.price:
            print(p.name, "is more expensive")
        else:
            print("Both products have the same price")


p1 = Product("Phone", 20000)
p2 = Product("Laptop", 50000)

p1.compare(p2)
```
