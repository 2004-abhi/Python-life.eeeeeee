```python
# all digit are evan
n= int(input("Enter a number: "))
all_even = True

while n != 0:
    digit = n % 10

    if digit % 2 != 0:
        all_even = False

    n = n // 10

if all_even:
    print("All digits are even")
else:
    print("Not all digits are even")
```

```python
# different b\t sum of evan and odd digits
n= int(input("Enter a number: "))
even_sum = 0
odd_sum = 0

while n != 0:
    digit = n % 10

    if digit % 2 == 0:
        even_sum += digit
    else:
        odd_sum += digit

    n = n // 10

difference = even_sum - odd_sum

print("Difference:", difference)
```

```python
# Check whether a number is prime
n = int(input("Enter a number: "))
i = 1
count = 0

while i <= n:
    if n % i == 0:
        count += 1

    i += 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
# Print all prime numbers from 1 to n
n = int(input("Enter n: "))
num = 2

while num <= n:
    i = 1
    count = 0

    while i <= num:
        if num % i == 0:
            count += 1
        i += 1

    if count == 2:
        print(num)

    num += 1
```

```python
# Count prime numbers from 1 to n
n = int(input("Enter n: "))

num = 2
prime_count = 0

while num <= n:
    i = 1
    count = 0

    while i <= num:
        if num % i == 0:
            count += 1
        i += 1

    if count == 2:
        prime_count += 1

    num += 1

print("Prime count:", prime_count)
```

```python
# Print Armstrong numbers from 1 to n
n = int(input("Enter n: "))

num = 1

while num <= n:
    temp = num
    sum = 0

    while temp != 0:
        digit = temp % 10
        sum += digit ** 3
        temp = temp // 10

    if sum == num:
        print(num)

    num += 1
    
```

```python
# Check whether a number is a strong number
n = int(input("Enter a number: "))

original = n
sum = 0

while n != 0:
    digit = n % 10

    factorial = 1
    i = 1

    while i <= digit:
        factorial *= i
        i += 1

    sum += factorial
    n = n // 10

if sum == original:
    print("Strong number")
else:
    print("Not a Strong number")
    
```

```python
# Print strong numbers within a range
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

num = start

while num <= end:
    original = num
    temp = num
    sum = 0

    while temp != 0:
        digit = temp % 10

        factorial = 1
        i = 1

        while i <= digit:
            factorial *= i
            i += 1

        sum += factorial
        temp = temp // 10

    if sum == original:
        print(num)

    num += 1
    
```
