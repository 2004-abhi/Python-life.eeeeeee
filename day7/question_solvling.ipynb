```python
# Take a year and determine whether it is a leap year using the correct 400/100/4 rules.
year = int(input("Enter a year: "))
if(year%400==0):
    print("IT a leap year ")
elif(year%100==0):
    print("not leap year ")
elif(year%4==0):
    print("leap year ")
else:
    print("no leap year ")
```

```python
# Take three sides and determine whether they can form a valid triangle. If valid, determine Equilateral, Isosceles, or Scalene.
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a == b == c:
    print("It is Equilateral")

elif a == b or a == c or b == c:
    print("It is Isosceles")

else:
    print("It is Scalene")
```
