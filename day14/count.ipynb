```python
# Count the frequency of each character in a word without using split().
word = input("Enter a word: ")

for i in range(len(word)):
    count = 0

    for j in range(len(word)):
        if word[i] == word[j]:
            count += 1

    print(word[i], "-", count)
```

```python
# Remove all duplicates from a list.
numbers = [10, 20, 10, 30, 20, 40, 30]

new_list = []

for n in numbers:
    if n not in new_list:   # checks whether the value of n is missing from the list new_list
        new_list += [n]      # This means add n to the list.
print(new_list)
```

```python
# Merge 2 lists without using methods/functions and arrange the merged list in ascending order.
list1 = [30, 10, 50]
list2 = [40, 20, 60]

new_list = []

for n in list1:
    new_list += [n]

for n in list2:
    new_list += [n]

for i in range(len(new_list)):
    for j in range(i + 1, len(new_list)):
        if new_list[i] > new_list[j]:
            temp = new_list[i]
            new_list[i] = new_list[j]
            new_list[j] = temp

print(new_list)
```

```python
# maximum difference of [7,4,1,2,3]
numbers = [7, 4, 1, 2, 3]
highest = numbers[0]
lowest = numbers[0]

for n in numbers:
    if n > highest:
        highest = n

    if n < lowest:
        lowest= n
    difference = highest - lowest
print("Maximum difference:", difference)
```

```python
# string expansion a3b4c1
text = input("Enter string: ")

result = ""
i = 0

while i < len(text):
    char = text[i]
    count = int(text[i + 1])

    result = result + char * count

    i += 2

print(result)
```

```python
# find the first non-repeating character in:swssip
word = input("Enter a word: ")

for i in range(len(word)):
    count = 0

    for j in range(len(word)):
        if word[i] == word[j]:
            count += 1

    if count == 1:
        print("First non-repeating character:", word[i])
        break
```

```python
a=[23,4,5,[6,7]]
b= a.copy()   # shallow copy
# Nested list is shared → changing nested data can affect both.
print(b)
a[3][1]=1000
print(a)
print(b)
```

```python
# Shallow Copy , We changed b: But because the inner list is shared, a also changes 
# Shallow → inner list is SHARED
a = [10, [20, 30]]

b = a.copy()

b[1][0] = 90

print("a:", a)
print("b:", b)
```

```python
import  copy
a=[10,[20,30]]
b=copy.deepcopy(a)    # creates a completely separate copy, including the nested list.
# Nested list is also copied → changing b does not affect a.
b[1][0]=90
print( "oriinal: ",a)
print( "deep copy :",b)
```

```python
# Deep Copy , Here only b changed , Because deepcopy() creates a completely separate copy, including the inner list.
# Deep    → inner list is SEPARATE
import copy

a = [10, [20, 30]]

b = copy.deepcopy(a)

b[1][0] = 90

print("a:", a)
print("b:", b)
```
