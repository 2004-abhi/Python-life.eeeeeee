```python
def decorator(fun):
    def inner():
        print("Before function")
        fun()
        print("After function")
    return inner

@decorator
def welcome():
    print("Welcome!")

welcome()
```

```python
def message(func): # func ---> add
    def inner():
        print("calculation started")
        func() # add()
        print("calculation completed")
    return inner
@message # take msg func as deco for this func add
def add():
    a=10
    b=20
    print("sum :",a+b)
add()   

    
```
