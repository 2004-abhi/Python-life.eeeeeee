```python
# super() is used to access methods of the parent class from a child class.
# We commonly use super() inside a child class's constructor to call the parent class's constructor.

# create two parent class and child class inherit using super()
class parent1:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class parent2:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class child(parent1,parent2):
    def __init__(self,c_name,c_age,name,age):
        super().__init__(name,age)
        self.c_name=c_name
        self.c_age=c_age
    def display(self):
        print("Name of father is :",self.name)
        print("name of age is :",self.age)
    
        print("name of child is :",self.c_name)
        print("name of age is :",self.c_age)

p1=child("Abhinand",22,"harilal",55)
p1.display()     

# In this example, super() calls only parent1's constructor.
# It does not automatically call parent2's constructor because these parent constructors aren't designed for cooperative multiple inheritance.
               
        
```

```python
class parent1:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class parent2:
    def __init__(self, mother_name, mother_age):
        self.mother_name = mother_name
        self.mother_age = mother_age

class child(parent1, parent2):
    def __init__(self, c_name, c_age, father_name, father_age, mother_name, mother_age):
        parent1.__init__(self, father_name, father_age)
        parent2.__init__(self, mother_name, mother_age)

        self.c_name = c_name
        self.c_age = c_age

    def display(self):
        print("Father Name:", self.name)
        print("Father Age:", self.age)
        print("Mother Name:", self.mother_name)
        print("Mother Age:", self.mother_age)
        print("Child Name:", self.c_name)
        print("Child Age:", self.c_age)

p1 = child("Abhinand", 22, "Harilal", 55, "Soumya", 48)
p1.display()
```

```python
class parent():
    def __init__(self,name):
        self.name=name

class child(parent):
    def __init__(self, name,c_name="Default"):
        super().__init__(name)
        self.c_name=c_name
        print("child name is :",self.c_name)
        print("parent name is :",self.name,self.c_name)
p1=parent("tom")
c1=child("Abu") 
c2=child("harilal","abhinand")      

# Your child class overrides __init__(), but it also uses super() to run the parent's constructor. 
# That is why the parent's name is still stored correctly.
        
```
