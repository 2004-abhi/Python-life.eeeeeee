```python
name = "counseling"
v = ""

# Step 1: Find the first vowel
for i in name:
    if i in "AEIOUaeiou":
        first_vowel = i
        break

# Step 2: Replace every vowel with the first vowel
for i in name:
    if i in "AEIOUaeiou":
        v += first_vowel
    else:
        v += i

print(v)
```

```python
num=[10,11,12,13,14,15]
evan_list=[]
odd_list=[]
for i in range (0,len(num)):
    if i % 2 == 0:
        evan_list.append(num[i]) 
    else:
        odd_list.append(num[i])
rev=odd_list[::-1]

print(evan_list)
print(odd_list)
print(rev)    

for i in range(0,len(evan_list)):
    print(evan_list[i],rev[i],end=" ")
```

```python
name = "Bibin"
score = 0

for i in range(len(name)):
    for j in range(i + 1, len(name) + 1):

        sub = name[i:j]

        if sub.lower() == sub[::-1].lower() and len(sub) > 1:
            print(sub, "Palindrome - 5 points")
            score += 5

print("Total Score:", score)
```
