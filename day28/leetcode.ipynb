```python
# 1. Two Sum – Find two indices whose elements add up to a target.
num=[11,33,15,45,12,2,7]
target=9
for i in range (len(num)):
    for j  in range (i+1, len(num)):
        if num[i] + num[j] == target :
            print(i,j)
    
```
