```python
# Accept a list from the user, find the sum of all values, then take the first 5 elements and print them in ascending order.

numbers = list(map(int, input("Enter numbers separated by space: ").split()))   # Take multiple numbers from the user and create a list of integers.
# map() gives the converted values, and list() makes them into an actual list.
# split() separates the text wherever there is a space.

total = sum(numbers)
print("Sum =", total)

first_five = numbers[:5]
first_five.sort()   # list[start : stop]
# .sort() arranges the list from smallest to largest.
print("First 5 elements in ascending order:", first_five) 
```

```python
# Accept 2 lists — one containing student names and one containing their marks. Find the lowest and highest marks along with the student's name, first using built-in methods and then without using methods.
names = ["Abhi", "Rahul", "Arun", "Vishnu"]
marks = [85, 72, 95, 60]
highest = marks[0]
lowest = marks[0]

highest_name = names[0]
lowest_name = names[0]

for i in range(len(marks)):
    if marks[i] > highest:
        highest = marks[i]
        highest_name = names[i]

    if marks[i] < lowest:
        lowest = marks[i]
        lowest_name = names[i]

print("Highest:", highest_name, highest)
print("Lowest:", lowest_name, lowest)
```

```python
# create a nestedlist and list full of string and check if the values are palindrome or not.
names = [["unnu","appa"
"", "hello"], ["level", "world", "radar","lallal","malayalam"]]

for group in names:
    for word in group:
        if word == word[::-1]:
            print(word, "is palindrome")
        else:
            print(word, "is not palindrome")
```

```python
# check a palindrome without slicing by manually creating the reverse.
word = input("Enter a word: ")

reverse = ""
i = len(word) - 1   # we're reversing the word, so we want to start from the last character.

while i >= 0:
    reverse = reverse + word[i]
    i -= 1

if word == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```
