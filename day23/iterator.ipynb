```python
# 1. Generator using yield and next()
def demo():
    yield 10
    yield 20

x = demo()

print(next(demo(), "Finished"))
print(next(demo(), "Finished"))

print(next(x, "Finished"))
print(next(x, "Finished"))
```

```python
# 2. Iterator using try-except
try:
    numbers = [10, 20]

    x = iter(numbers)

    print(next(x))
    print(next(x))
    print(next(x))

except StopIteration as e:
    print(e, "StopIteration")
```

```python
# create a generator function that generates numbers from 10 to 50 with a difference of 10
def generator():
    for i in range (10,51,10):
        yield i # produce one number at a time
        # yield is used in generator function to produce values one by one

for i in generator():
    print (i)        

```

```python
#  create a generator function that generates the multiplication table of 5
def generator():
    for i in range (5,51,5):
        yield i
n=1
for i in  generator():
    print(n," x 5 =",i )
    n=n+1
        
```

```python
# create a generator function that generates squares of numbers 1 from to 5
def generator():
    for i in range (1,6):
        yield i * i
for i in generator():
    print(i)
```

```python
# create a generator function that generate numbers divisible by  5  b/w 1 to 50
def num():
    for i in range (1 , 51):
        if i %5==0:
            yield i
for i in num():
    print(i)             
```
