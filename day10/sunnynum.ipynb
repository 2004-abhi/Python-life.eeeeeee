```python
# the no is automorphic no or not //An Automorphic Number is a number whose square ends with the same number.
n = int(input("Enter a number: "))

square = n * n
temp = n

while temp > 0:
    if square % 10 != temp % 10:
        print("Not an Automorphic Number")
        break

    square = square // 10
    temp = temp // 10

else:
    print("Automorphic Number")
```
