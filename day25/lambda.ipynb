```python
# lambda
# A lambda function is a small anonymous function (a function without a name) that can contain only one expression.
# In simple words, we use lambda to write a function in a single line.
# lambda arguments: expression (syntax)
add = lambda a, b: a + b
print(add(10, 20))
```

```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(10))
print(check(7))
```

```python
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * 2, numbers))

print(result)
```
