```python
# 1. What is List Comprehension?
# List comprehension is a short and simple way to create a new list using a single line of code.
# We can use it instead of writing a normal for loop
# 3. Syntax
# new_list = [expression for item in iterable]

numbers = [14, 23, 13, 4, 5]

result = [i for i in numbers if i>10]

print(result)
```

```python
num=[10,20,30,40,50]
l=[i*2 for i in num]
print(l)
```

```python
numbers = [[1, 2], [3, 4], [5, 6]]

result = [j for i in numbers for j in i]

print(result)
```

```python
result = [(i, j) for i in range(1, 3) for j in range(1, 4)]

print(result)
```

```python
l = [(i, j) for i in [1, 2] for j in [10, 20]]

print(l)

# Step-by-step:
# 1. First, i = 1.
# 2. Then j = 10 → (1, 10).
# 3. Next j = 20 → (1, 20).
# 4. Now the inner loop finishes.
# 5. Next i = 2.
# 6. Again j = 10 → (2, 10).
# 7. Next j = 20 → (2, 20).

# Iteration	i	       j	         Result
# 1	        1	       10	        (1, 10)
# 2	        1	       20	        (1, 20)
# 3	        2	       10	        (2, 10)
# 4	        2	       20	        (2, 20)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Student data
data = {
    "Name": ["Abhi", "Arun", "Rahul", "Anu", "John"],
    "Python": [85, 70, 90, 95, 60],
    "SQL": [80, 75, 85, 90, 65],
    "Java": [90, 80, 88, 92, 70]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate total marks
df["Total"] = df[["Python", "SQL", "Java"]].sum(axis=1)

# Calculate average using NumPy
df["Average"] = np.mean(
    df[["Python", "SQL", "Java"]], axis=1
)

# Display student details
print(df)

# Find topper
topper = df.loc[df["Total"].idxmax()]
print("\nTopper:", topper["Name"])
print("Highest Total:", topper["Total"])

# Bar chart using Matplotlib
plt.bar(df["Name"], df["Total"])
plt.xlabel("Students")
plt.ylabel("Total Marks")
plt.title("Student Total Marks")
plt.show()

# Heatmap using Seaborn
sns.heatmap(
    df[["Python", "SQL", "Java"]],
    annot=True,
    cmap="coolwarm",
    xticklabels=True,
    yticklabels=df["Name"]
)

plt.title("Subject Marks Heatmap")
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sales data
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [15000, 20000, 18000, 25000, 30000, 22000],
    "Expenses": [5000, 7000, 6000, 8000, 10000, 7500]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate profit
df["Profit"] = df["Sales"] - df["Expenses"]

# Calculate total sales using NumPy
total_sales = np.sum(df["Sales"])

# Calculate average sales
average_sales = np.mean(df["Sales"])

# Display data
print("SALES REPORT")
print(df)

print("\nTotal Sales:", total_sales)
print("Average Sales:", average_sales)

# Find highest sales month
highest = df.loc[df["Sales"].idxmax()]

print("\nHighest Sales Month:", highest["Month"])
print("Highest Sales:", highest["Sales"])

# Matplotlib - Sales graph
plt.plot(df["Month"], df["Sales"], marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Analysis")
plt.show()

# Seaborn - Profit bar chart
sns.barplot(data=df, x="Month", y="Profit")
plt.title("Monthly Profit Analysis")
plt.show()

# Seaborn - Correlation heatmap
sns.heatmap(
    df[["Sales", "Expenses", "Profit"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Sales Correlation")
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Weather data
data = {
    "Day": ["Monday", "Tuesday", "Wednesday",
            "Thursday", "Friday", "Saturday", "Sunday"],
    "Temperature": [30, 32, 29, 35, 33, 28, 31],
    "Humidity": [65, 70, 60, 75, 68, 55, 62],
    "Rainfall": [0, 5, 0, 10, 2, 0, 3]
}

# Create DataFrame
df = pd.DataFrame(data)

print("WEATHER REPORT")
print(df)

# NumPy calculations
avg_temp = np.mean(df["Temperature"])
max_temp = np.max(df["Temperature"])
min_temp = np.min(df["Temperature"])

print("\nAverage Temperature:", avg_temp)
print("Maximum Temperature:", max_temp)
print("Minimum Temperature:", min_temp)

# Find hottest day
hottest = df.loc[df["Temperature"].idxmax()]
print("\nHottest Day:", hottest["Day"])

# Find coldest day
coldest = df.loc[df["Temperature"].idxmin()]
print("Coldest Day:", coldest["Day"])

# Pandas filtering
print("\nDays with temperature above 30:")
print(df[df["Temperature"] > 30])

# Matplotlib - Temperature graph
plt.plot(df["Day"], df["Temperature"], marker="o")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Analysis")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Seaborn - Humidity chart
sns.barplot(data=df, x="Day", y="Humidity")
plt.title("Weekly Humidity Analysis")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Seaborn - Weather heatmap
sns.heatmap(
    df[["Temperature", "Humidity", "Rainfall"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Weather Data Correlation")
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Product data
data = {
    "Product": ["Laptop", "Mobile", "Headphone", "Keyboard", "Mouse", "Monitor"],
    "Price": [50000, 20000, 2000, 1500, 800, 12000],
    "Quantity": [5, 10, 20, 15, 30, 8],
    "Rating": [4.5, 4.2, 4.0, 3.8, 4.3, 4.6]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate revenue
df["Revenue"] = df["Price"] * df["Quantity"]

print("PRODUCT DETAILS")
print(df)

# NumPy calculations
print("\nTotal Revenue:", np.sum(df["Revenue"]))
print("Average Price:", np.mean(df["Price"]))
print("Highest Price:", np.max(df["Price"]))

# Find best-selling product by quantity
best_selling = df.loc[df["Quantity"].idxmax()]

print("\nBest Selling Product:", best_selling["Product"])
print("Quantity Sold:", best_selling["Quantity"])

# Find highest revenue product
top_revenue = df.loc[df["Revenue"].idxmax()]

print("\nHighest Revenue Product:", top_revenue["Product"])

# Pandas filtering
print("\nProducts with rating above 4:")
print(df[df["Rating"] > 4])

# Matplotlib - Revenue chart
plt.bar(df["Product"], df["Revenue"])
plt.xlabel("Products")
plt.ylabel("Revenue")
plt.title("Product Revenue Analysis")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Seaborn - Product rating chart
sns.barplot(data=df, x="Product", y="Rating")
plt.title("Product Rating Analysis")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Seaborn - Correlation heatmap
sns.heatmap(
    df[["Price", "Quantity", "Rating", "Revenue"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Product Data Correlation")
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Employee data
data = {
    "Name": ["Abhi", "Arun", "Rahul", "Anu", "John", "Meera"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Salary": [35000, 25000, 45000, 40000, 30000, 50000],
    "Experience": [2, 1, 4, 3, 2, 5]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate annual salary
df["Annual Salary"] = df["Salary"] * 12

print("EMPLOYEE DETAILS")
print(df)

# NumPy calculations
print("\nAverage Salary:", np.mean(df["Salary"]))
print("Maximum Salary:", np.max(df["Salary"]))
print("Minimum Salary:", np.min(df["Salary"]))

# Find highest-paid employee
highest = df.loc[df["Salary"].idxmax()]

print("\nHighest Paid Employee:", highest["Name"])
print("Salary:", highest["Salary"])

# Department-wise average salary
print("\nDepartment-wise Average Salary:")
print(df.groupby("Department")["Salary"].mean())

# Filter employees with salary above 30000
print("\nEmployees with salary above 30000:")
print(df[df["Salary"] > 30000])

# Matplotlib - Employee salary chart
plt.bar(df["Name"], df["Salary"])
plt.xlabel("Employee")
plt.ylabel("Salary")
plt.title("Employee Salary Analysis")
plt.show()

# Seaborn - Department salary comparison
sns.barplot(data=df, x="Department", y="Salary")
plt.title("Department-wise Salary")
plt.show()

# Seaborn - Experience vs Salary
sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary",
    hue="Department",
    s=100
)
plt.title("Experience vs Salary")
plt.show()

```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Movie dataset
data = {
    "Movie": ["Leo", "Jailer", "Manjummel Boys",
              "Premalu", "Aavesham", "2018", "Drishyam"],
    "Genre": ["Action", "Action", "Drama",
              "Comedy", "Comedy", "Drama", "Thriller"],
    "Rating": [7.8, 7.2, 8.2, 7.9, 7.8, 8.0, 8.3],
    "Collection": [600, 650, 200, 136, 156, 175, 75]
}

# Create DataFrame
df = pd.DataFrame(data)

print("MOVIE DETAILS")
print(df)

# NumPy calculations
print("\nAverage Rating:", np.mean(df["Rating"]))
print("Highest Rating:", np.max(df["Rating"]))
print("Lowest Rating:", np.min(df["Rating"]))

# Find highest-rated movie
top_movie = df.loc[df["Rating"].idxmax()]

print("\nHighest Rated Movie:", top_movie["Movie"])
print("Rating:", top_movie["Rating"])

# Find movies with rating above 8
print("\nMovies with rating above 8:")
print(df[df["Rating"] > 8])

# Genre-wise average rating
print("\nGenre-wise Average Rating:")
print(df.groupby("Genre")["Rating"].mean())

# Matplotlib - Movie rating chart
plt.bar(df["Movie"], df["Rating"])
plt.xlabel("Movies")
plt.ylabel("Rating")
plt.title("Movie Rating Analysis")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Seaborn - Genre-wise rating
sns.barplot(data=df, x="Genre", y="Rating")
plt.title("Genre-wise Average Rating")
plt.show()

# Seaborn - Collection vs Rating
sns.scatterplot(
    data=df,
    x="Rating",
    y="Collection",
    hue="Genre",
    s=100
)
plt.title("Movie Rating vs Collection")
plt.show()
```

```python
class Father:
    def __init__(self):
        self.name = "Arun"
        self.age = 50

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


class Son(Father):
    def show(self):
        print("I am the son")


s1 = Son()

s1.display()
s1.show()
```
