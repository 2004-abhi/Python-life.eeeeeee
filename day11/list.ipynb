```python
list_name=[10,20,30,40,50,80,100,90]
# build in function
# 1. len()-n elements in the list
print(len(list_name))
print(sum(list_name))
print(max(list_name))
print(min(list_name))
print(sorted(list_name))
print(sorted(list_name , reverse=True))
print(all(list_name))
print(any(list_name))
```

```python
list_namee = [10, 20, 30, 40, 50, 80, 100, 90]

# Adding an element using append
list_namee.append(101)

print(list_namee)

# nesrsted append
list_namee.append([12])
print(list_namee)
```

```python
list_namee=[10,20,30,40,50,80,100,90]
list_namee.insert(2,20000)
print(list_namee)
# to insert element into which index position
```

```python
list_namee=[10,20,30,40,50,80,100,90]
list_namee.extend([101,103,104])
print(list_namee)
#  it extend/add the values to the last of the current list
```

```python
list_namee=[10,20,30,40,50,80,100,90]
list_namee.pop(1)
print(list_namee)
#  to pop the index value of 2 from the list
```

```python
list_namee=[10,20,30,40,50,80,100,90]
# index returns a value , find the index value of that element
print(list_namee.index(20))
print(list_namee)
```
