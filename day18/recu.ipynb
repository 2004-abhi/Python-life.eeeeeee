```python
# 1. Print numbers from 1 to 10
def show(i):
    if i <= 10:
        print(i)
        show(i + 1)

show(1)
```

```python
# 2. Print only even numbers from 0 to 10
def show(i):
    if i <= 10:
        if i % 2 == 0:
            print(i)
        show(i + 1)

show(0)
```

```python
# 3. Print only odd numbers from 1 to 9
def show(i):
    if i <= 9:
        if i % 2 != 0:
            print(i)
        show(i + 1)

show(1)
```

```python
# 4. Print the number 10 five times
def show(i):
    if i < 5:
        print(10)
        show(i + 1)

show(0)
```

```python
# 5. Print numbers only while i < 10
def show(i):
    if i < 10:
        print(i)
        show(i + 1)

show(1)
```

```python
# 6. Print a number only if the number is positive
def show(i):
    if i > 0:
        print(i)
        show(i - 1)

show(5)
```

```python
# 7. Print numbers from 1 to 20 only when the number is less than 15
def show(i):
    if i <= 20:
        if i < 15:
            print(i)
        show(i + 1)

show(1)
```

```python
# 8. Print numbers from 1 to 20 only when the number is greater than 10
def show(i):
    if i <= 20:
        if i > 10:
            print(i)
        show(i + 1)

show(1)
```

```python
# 9. Print numbers from 1 to 10 only when the number is not equal to 5
def show(i):
    if i <= 10:
        if i != 5:
            print(i)
        show(i + 1)

show(1)
```

```python
# 10. Print numbers from 1 to 10 only when the number is divisible by 2
def show(i):
    if i <= 10:
        if i % 2 == 0:
            print(i)
        show(i + 1)

show(1)
```
