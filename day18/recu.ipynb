```python
for i in range(1,6):
    for j in range(1,i):   # i changes every time.
        print(j,end=" ")
    print()    
# | `i` | `range(1, i)` | Values of `j` |
# |---:|---|---|
# | 1 | `range(1,1)` | Nothing |
# | 2 | `range(1,2)` | 1 |
# | 3 | `range(1,3)` | 1, 2 |
# | 4 | `range(1,4)` | 1, 2, 3 |
# | 5 | `range(1,5)` | 1, 2, 3, 4 |
```
