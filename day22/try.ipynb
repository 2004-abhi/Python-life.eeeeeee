```python
# 1. Define the custom exception
class InsufficientBalanceError(Exception):
    """Exception raised when a withdrawal amount exceeds the available balance."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        self.shortfall = amount - balance
        super().__init__(f"Attempted to withdraw ${amount}, but balance is only ${balance}. Shortfall: ${self.shortfall}")

# 2. Implement the Bank Account class
class BankAccount:
    def __init__(self, initial_balance):
        self.balance = initial_balance

    def withdraw(self, amount):
        if amount > self.balance:
            # Raise the custom exception if funds are lacking
            raise InsufficientBalanceError(self.balance, amount)
        
        self.balance -= amount
        print(self.balance)
        return self.balance

# 3. Test the exception handling
account = BankAccount(initial_balance=500)

try:
    print("Attempting to withdraw $200...")
    account.withdraw(200)
    
    print("\nAttempting to withdraw $400...")
    account.withdraw(400) # This will trigger the exception
    
except InsufficientBalanceError as e:
    print(f"Transaction Failed: {e}")
```

```python
# find the factorial of a numbers without using **
def multiply(a,b):
    if b==0:
        return 0
    return a + multiply(a,b-1)

def fact(n):
    if n==0 or n==1:
        return 1
    return multiply(n, fact(n-1))
n=int(input("enter a number :"))
print("Factorial = ",fact(n))
```

```python
# 1. Find factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


print(factorial(5))
```

```python
# 2. Reverse a string using recursion
def reverse(s):
    if s == "":
        return s

    return reverse(s[1:]) + s[0]


print(reverse("hello"))
```

```python
# 3. Find the sum of digits using recursion
def sum_digits(n):
    if n == 0:
        return 0

    return n % 10 + sum_digits(n // 10)
# % 10 → Gets the last digit.
# // 10 → Removes the last digit.

print(sum_digits(1234))
```

```python
# 4. Check whether a number is prime using a function
n = int(input("Enter number: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count += 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
def prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


n = int(input("Enter number: "))

if prime(n):
    print("Prime number")
else:
    print("Not a prime number")
```

```python
# 7. Create a generator that produces even numbers up to N
def even(n):
    for i in range(2, n + 1, 2):
        yield i


for i in even(10):
    print(i)
```

```python
# 8. Create a generator for Fibonacci numbers
def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        yield a

        a, b = b, a + b
# The Fibonacci series is a sequence of numbers in which each new number is the sum of the previous two numbers.

for i in fibonacci(7):
    print(i)
```

```python
# 9. Recursive function to find the largest element in a list
def largest(a):
    if len(a) == 1:
        return a[0]

    max_value = largest(a[1:])

    if a[0] > max_value:
        return a[0]
    else:
        return max_value


a = [10, 50, 20, 80, 30]

print("Largest:", largest(a))
```
