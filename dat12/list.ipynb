```python
# crate a list and count no of palindrome in list
number=[121,333,245,34543,7545]
count =0
for n in number:  
    # Take each number from the list one by one and store it in n
    ori=n
    # This creates a copy of the current number.
    reverse=0
    # We need a variable to build the reversed number.

    while n!=0:
        # Keep repeating the loop until n becomes 0.
        digit = n%10 
        # % 10 gives us the last digit.   
        reverse=reverse*10+digit
        # Because we need to move the existing number one place to the left.
        n=n//10
        # // 10 removes the last digit.

    if ori==reverse:
        print(ori,"is palindrome ")
        count +=1
print("number of palindrome:",count)        

```

```python
# crate a list full of palindrome and find the shortest as well as longest
number=[121,333,245,34543,7545]
# [0] means first element of the list.
shortest = number[0]
longest = number[0]
for n in number:
    if len(str(n))<len(str(shortest)):             
        shortest=n
    if len(str(n))>len(str(longest)):
        longest=n
print("Shortest palindrome:", shortest)
print("Longest palindrome:", longest)            
```

```python
# What is a Buzz Number?
# A number is a Buzz Number if:
# It is divisible by 7, OR
# Its last digit is 7.
n = int(input("Enter a number: "))

if n % 7 == 0 or n % 10 == 7:
    print("Buzz Number")
else:
    print("Not a Buzz Number")
```
